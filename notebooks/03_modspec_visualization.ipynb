# 03 — Visualización de modulation spectrums

- Plot de modspec STFT vs CWT para un mismo epoch.
- Promedio por clase (AD vs HC) en ambas variantes.
- Comparación visual con la Figura 4 del paper de Lopes.

```python
%load_ext autoreload
%autoreload 2
import sys; sys.path.append('..')
import numpy as np
import matplotlib.pyplot as plt
from src.modspec import ModSpecConfig, compute_modulation_spectrum
from src.utils.viz import plot_modspec, annotate_canonical_bands

rng = np.random.default_rng(0)
fs = 200
t = np.arange(8 * fs) / fs
x = np.stack([np.sin(2*np.pi*10*t) + 0.5*np.sin(2*np.pi*4*t)*np.cos(2*np.pi*1*t) + 0.1*rng.standard_normal(t.size) for _ in range(19)])
M_stft = compute_modulation_spectrum(x, ModSpecConfig(method='stft', fs=fs))
M_cwt = compute_modulation_spectrum(x, ModSpecConfig(method='cwt', fs=fs))
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
plot_modspec(M_stft.mean(0), title='STFT — promedio canales', ax=axes[0]); annotate_canonical_bands(axes[0])
plot_modspec(M_cwt.mean(0), title='CWT — promedio canales', ax=axes[1]); annotate_canonical_bands(axes[1])
plt.tight_layout();
```
