- Filtro 0.5–45 Hz: verificar atenuación fuera de banda en PSD.

- Pico alpha en O1/O2 esperado en HC.
- Conteo de epochs por sujeto ≥460.

```python
%load_ext autoreload
%autoreload 2
import sys; sys.path.append('..')
from pathlib import Path
import mne
from src.io_bids import list_subjects, load_raw
from src.preprocess import PreprocessConfig, preprocess_raw
from src.epoching import make_epochs

subs = list_subjects('../data/raw/ds004504', groups_keep=('A', 'C'))
rec = subs[0]
raw = load_raw(rec)
raw_pp = preprocess_raw(raw, PreprocessConfig(channels_keep=tuple(['Fp1','Fp2','F7','F3','Fz','F4','F8','T3','C3','Cz','C4','T4','T5','P3','Pz','P4','T6','O1','O2'])))
raw_pp.compute_psd(fmax=50).plot();
```
